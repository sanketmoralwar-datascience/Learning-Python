## 1. What is a Function?

A function is a reusable block of code that performs a particular task.

Instead of writing the same code repeatedly, we can write it once inside a function and execute it whenever required.

### Example: Without a function

Suppose you want to print a welcome message three times.

```
print("Welcome to Python!")print("Welcome to Python!")print("Welcome to Python!")
```

Here, we are repeating the same code three times.

### Example: With a function

```
def welcome():    print("Welcome to Python!")
```

Now, whenever you want to execute this code, simply call the function:

```
welcome()welcome()welcome()
```

Output:

```
Welcome to Python!
Welcome to Python!
Welcome to Python!
```

Notice that we wrote the logic only once but reused it three times.

That's the main advantage of functions!

## 2. Why do we use Functions?

Functions make programming easier and more organized.

Some important advantages:

1. Code Reusability: Write code once and use it multiple times.
2. Less Repetition: Avoid writing identical code repeatedly.
3. Easy Maintenance: Modify the code in one place.
4. Better Organization: Divide large programs into smaller tasks.
5. Readability: Make programs easier to understand.

For example, in a student management program, we could create separate functions for:

- Calculating student percentage.
- Checking pass or fail.
- Displaying student details.
- Calculating grades.

## 3. How to Create a Function

Python uses the `def` keyword to define a function.

Syntax:

```
def function_name():    # Statements
```

Let's understand the different parts:

| Component       | Meaning                                          |
| --------------- | ------------------------------------------------ |
| `def`           | Keyword used to define a function                |
| `function_name` | Name given to the function                       |
| `()`            | Parentheses used for parameters                  |
| `:`             | Indicates the beginning of the function body     |
| Indentation     | Defines the statements belonging to the function |

### Example

```
def greet():    print("Hello Sanket!")
```

Important: Defining a function does not automatically execute it.

You must call the function.

```
greet()
```

Output:

```
Hello Sanket!
```

## 4. Function Parameters and Arguments

Functions can accept information from outside.

For example, instead of creating separate functions for greeting different people, we can create one function that accepts a name.

### Example

```
def greet(name):    print("Hello", name)
```

Calling the function:

```
greet("Sanket")greet("Rahul")greet("Amit")
```

Output:

```
Hello Sanket
Hello Rahul
Hello Amit
```

Let's understand two important terms:

- Parameter: A variable mentioned inside the function definition.
- Argument: The actual value passed when calling the function.

In the above example:

```
def greet(name):
```

`name` is a parameter.

```
greet("Sanket")
```

`"Sanket"` is an argument.

## 5. Functions with Multiple Parameters

A function can accept multiple parameters.

### Example

```
def student_details(name, age, subject):    print("Student Name:", name)    print("Age:", age)    print("Subject:", subject)
```

Calling the function:

```
student_details("Sanket", 25, "Python")
```

Output:

```
Student Name: Sanket
Age: 25
Subject: Python
```

Here, the function accepts three parameters:

- `name`
- `age`
- `subject`

You can pass different values whenever you call the function.

## 6. The return Statement

One of the most important concepts in Python functions is `return`.

The `return` statement sends a result back from a function to the place where it was called.

### Example: Using print()

```
def addition(a, b):    print(a + b)addition(10, 20)
```

Output:

```
30
```

This displays the result, but it doesn't give the result back to the calling code.

### Example: Using return

```
def addition(a, b):    return a + b
```

Now we can store the returned result in a variable.

```
result = addition(10, 20)print(result)
```

Output:

```
30
```

We can also use the returned value in further calculations.

```
result = addition(10, 20)final_result = result * 2print(final_result)
```

Output:

```
60
```

### Difference between print() and return

| print()                             | return                              |
| ----------------------------------- | ----------------------------------- |
| Displays information on the screen  | Sends a value back to the caller    |
| Mainly used for displaying output   | Used when we need to reuse a result |
| Does not return the displayed value | Returns the specified value         |

Remember: `return` is particularly useful when building real-world programs.

## 7. Default Parameters

Sometimes we want a function to use a default value when the user doesn't provide an argument.

Python allows this using default parameters.

### Example

```
def greet(name="Guest"):    print("Welcome", name)
```

Calling the function:

```
greet()greet("Sanket")
```

Output:

```
Welcome Guest
Welcome Sanket
```

Explanation:

- When no argument is provided, Python uses `"Guest"`.
- When an argument is provided, Python uses that value instead.

## 8. Keyword Arguments

Normally, Python assigns arguments according to their position.

However, we can also pass arguments using parameter names.

### Example

```
def student(name, age, city):    print("Name:", name)    print("Age:", age)    print("City:", city)
```

Calling using keyword arguments:

```
student(city="Mumbai", name="Sanket", age=25)
```

Output:

```
Name: Sanket
Age: 25
City: Mumbai
```

Notice that the order of arguments doesn't matter when we explicitly specify parameter names.

## 9. Variable-Length Arguments (\*args)

Sometimes we don't know how many arguments will be passed to a function.

Python provides `*args` for this purpose.

### Example

```
def calculate_sum(*numbers):    total = 0    for num in numbers:        total = total + num    return total
```

Calling the function:

```
print(calculate_sum(10, 20))print(calculate_sum(10, 20, 30))print(calculate_sum(10, 20, 30, 40))
```

Output:

```
30
60
100
```

Here:

- `*args` accepts multiple positional arguments.
- Python collects them into a tuple.
- We can iterate through that tuple using a loop.

## 10. Variable-Length Keyword Arguments (\*\*kwargs)

Similar to `*args`, Python provides `**kwargs` to accept multiple keyword arguments.

These arguments are collected into a dictionary.

### Example

```
def student_details(**details):    for key, value in details.items():        print(key, ":", value)
```

Calling the function:

```
student_details(    name="Sanket",    age=25,    subject="Python",    city="Mumbai")
```

Output:

```
name : Sanket
age : 25
subject : Python
city : Mumbai
```

Remember:

| Feature    | Purpose                       | Stored as  |
| ---------- | ----------------------------- | ---------- |
| `*args`    | Multiple positional arguments | Tuple      |
| `**kwargs` | Multiple keyword arguments    | Dictionary |

## 11. Scope of Variables

Variable scope determines where a variable can be accessed.

There are two important types to understand initially.

### A. Local Variable

A variable created inside a function is called a local variable.

```
def my_function():    message = "Hello Python"    print(message)my_function()
```

Output:

```
Hello Python
```

The variable `message` belongs to the function.

### B. Global Variable

A variable created outside a function is called a global variable.

```
message = "Welcome to Python"def my_function():    print(message)my_function()
```

Output:

```
Welcome to Python
```

A function can access a global variable.

For now, remember:

- Local variables are created inside functions.
- Global variables are created outside functions.

## 12. Important Rules to Remember

1. Every function definition starts with `def`.
2. A function executes only when it is called.
3. Function names should be meaningful.
4. Function bodies must be properly indented.
5. Functions can accept zero or more parameters.
6. Functions can return values using `return`.
7. A function without an explicit return statement returns `None`.
8. You can define one function and call it multiple times.
