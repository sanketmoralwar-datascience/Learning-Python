#random - Flipping a coin

```python
import random

coin = random.choice(["heads", "tails"])
print(coin)
```

same as above code

```python
from random import choice

coin = choice(["heads", "tails"])
print(coin)
```

# Application Programming Interface (API)
requests

```python
# pip install requests
import requests
import json

response = requests.get("https://itunes.apple.com/search?entity=song&limit=1&term=weezer")
print(json.dumps(response.json(), indent=2))
```

```python
response = requests.get("https://itunes.apple.com/search?entity=song&limit=50&term=weezer") #Changing limit to 50 from 1
#print(json.dumps(response.json(), indent=2))

o = response.json()
for result in o["results"]:
  print(result["trackName"])
```
