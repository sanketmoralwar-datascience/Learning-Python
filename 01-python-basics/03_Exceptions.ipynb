ValueError

```python
x = int(input("What's x? "))
print(f"x is {x}")
```

```python
try:
  x = int(input("What's x? "))
  print(f"x is {x}")
except ValueError:
  print("x is not an integer")
```

NameError

```python
try:
  x = int(input("What's x? "))
except ValueError:
  print("x is not an integer")

print(f"x is {x}")  # we are getting NameError because the input function which assigns value to x has error and x remains undefined.
```

```python
try:
  x = int(input("What's x? "))
except ValueError:
  print("x is not an integer")
else:
  print(f"x is {x}")
```

```python
while True:
    try:
      x = int(input("What's x? "))
    except ValueError:
      print("x is not an integer")
    else:
      break

print(f"x is {x}")
```

```python
def main():
  x = get_int()
  print(f"x is {x}")


def get_int():
  while True:
    try:
      x = int(input("What's x? "))
    except ValueError:
      print("x is not an integer")
    else:
      return x

main()
```

Another way is to use pass, so that we dont bombard with same message again and again

```python
def main():
  x = get_int()
  print(f"x is {x}")


def get_int():
  while True:
    try:
      return int(input("What's x? "))
    except ValueError:
      pass


main()
```

prompt : for reusable function. instead if whats x we can call it whats y

```python
def main():
  x = get_int("Whats x")
  print(f"x is {x}")

  y = get_int("Whats y?")
  print(f"y is {y}")


def get_int(prompt):
  while True:
    try:
      return int(input(prompt))
    except ValueError:
      pass


main()
```
