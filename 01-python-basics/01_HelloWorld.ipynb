"#" is used to add comments in the code \
Different ways to use print statement

```python
# Ask user for their name
name = input("What's your Name? ")

# Say Hello to the user

#Type 1:
print("hello, "+ name)

#Type 2:
print("hello,", name)

#Type 3: Default value for end is"\n" we can override this value with blank
print("hello, ", end='')
print(name)

#Type 4 : f string
print(f"hello, {name}")

```

functions for type str

```python
print(f"hello, {name} ")

name = name.strip()
print(f"hello, {name} ")

name = name.capitalize()
print(f"hello, {name} ")

name = name.title()
print(f"hello, {name} ")

name = name.upper()
print(f"hello, {name} ")

name = input("What do you do?").strip().lower()
print(f"hello, {name} ")

```

functions for type int  

```python
x = 1
y = 2

z = x + y

print(z)

x = input("What's x? ")
y = input("What's y? ")
z = (x) + (y)
print(z)

a = int(x) + int(y)
print(a)
```

Removing extra code

```python
x = int(input("What's x? "))
y = int(input("What's y? "))

print(x + y)

#More compact but less readable
print(int(input("What's x? ")) + int(input("What's y? ")))
```

functions for type float

```python
x = float(input("What's x? "))
y = float(input("What's y? "))

z = x + y
print(z)

print(round(z))

#formated
print(f"{z:,}")


# Round with 2 places after decimal
a = x / y
print(round(a,2))
print(f"{a:.2f}")
```
