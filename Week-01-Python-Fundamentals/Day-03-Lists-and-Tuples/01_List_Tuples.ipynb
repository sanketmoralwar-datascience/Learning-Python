# Part 1 – Introduction to Lists

## 1. What is a List in Python?

A list is a collection of multiple items stored inside a single variable.

For example, imagine you want to store the marks of five students.

Without a list, you might write:

```
mark1 = 85
mark2 = 90
mark3 = 78
mark4 = 92
mark5 = 88
```

This works, but imagine storing marks for 100 students!

It would become difficult to manage.

Instead, Python provides lists.

```
marks = [85, 90, 78, 92, 88]

print(marks)
```

Output:

```
[85, 90, 78, 92, 88]
```

Now all five marks are stored in one variable.

### Important characteristics of Lists

1. Ordered: Items maintain their position.
2. Mutable: You can modify items after creating a list.
3. Allow duplicates: The same value can appear multiple times.
4. Different data types: A list can contain integers, strings, floats, and Boolean values.
5. Indexed: Each item has an index starting from zero.

## 2. Creating Lists

You create a list using square brackets `[]`.

Example 1 – List of integers

```
numbers = [10, 20, 30, 40, 50]
print(numbers)
```

Output:


Example 2 – List of strings

```
subjects = ["Python", "SQL", "Statistics", "Excel"]
print(subjects)
```

Output:

```

```

Example 3 – Mixed data types

```
student = ["Sanket", 25, 85.5, True]
print(student)
```

Output:

```
['Sanket', 25, 85.5, True]
```

Example 4 – Empty list

```

print(my_list)
```

```
[]
```

You can also check the data type using `type()`.

```
numbers = [10, 20, 30]

```

Output:

```
<class 'list'>
```

# Part 2 – Accessing List Elements

## 3. Positive Indexing

Every element in a list has an index.

Python starts counting from 0, not 1.

Consider this list:

```
languages = ["Python", "Java", "C++", "SQL", "R"]
```

The index positions are:

| Element | Positive Index |
| ------- | -------------- |
| Python  | 0              |
| Java    | 1              |
| C++     | 2              |
| SQL     | 3              |
| R       | 4              |

Example 1 – Access the first element

```
languages = ["Python", "Java", "C++", "SQL", "R"]
print(languages[0])
```

Output:

```
Python
```

Example 2 – Access the third element

```
print(languages[2])
```

Output:

```
C++
```

Example 3 – Access the last element

```
print(languages[4])
```

Output:

```
R
```

## 4. Negative Indexing

Python also supports negative indexing.

Negative indexing starts counting from the end of the list.

| Element | Positive Index | Negative Index |
| ------- | -------------- | -------------- |
| Python  | 0              | -5             |
| Java    | 1              | -4             |
| C++     | 2              | -3             |
| SQL     | 3              | -2             |
| R       | 4              | -1             |

Example 1 – Access the last element

```
languages = ["Python", "Java", "C++", "SQL", "R"]
print(languages[-1])
```

Output:

```
R
```

Example 2 – Access the second-last element

```
print(languages[-2])
```

Output:

```
SQL
```

Remember:

- Positive indexing starts from the left.
- Negative indexing starts from the right.
- `0` represents the first element.
- `-1` represents the last element.

# Part 3 – List Slicing

## 5. What is List Slicing?

Slicing allows you to extract multiple elements from a list.

The syntax is:

```
list_name[start:stop:step]
```

Where:

- `start`: Starting index (included).
- `stop`: Ending index (excluded).
- `step`: Number of positions to move.

Let's understand with examples.

Example 1 – Basic slicing

```
numbers = [10, 20, 30, 40, 50, 60]
print(numbers[1:4])

```

Output:

```
[20, 30, 40]
```

Why?

- Starts at index 1 → 20
- Stops before index 4 → 50 is excluded.

Example 2 – Slice from the beginning

```
print(numbers[:3])
```

Output:

```
[10, 20, 30]
```

Example 3 – Slice until the end

```
print(numbers[2:])
```

Output:

```
[30, 40, 50, 60]
```

Example 4 – Using step

```
print(numbers[::2])
```

Output:

```
[10, 30, 50]
```

Here, Python selects every second element.

Example 5 – Reverse a list using slicing

```
print(numbers[::-1])
```

Output:

```
[60, 50, 40, 30, 20, 10]
```

This is a very useful technique in Python.

# Part 4 – Modifying Lists

## 6. Lists are Mutable

Mutable means you can change the contents of a list after creating it.

Example – Changing an element

```
fruits = ["Apple", "Banana", "Mango"]
fruits[1] = "Orange"
print(fruits)
```

Output:

```
['Apple', 'Orange', 'Mango']
```

Notice that Banana was replaced with Orange.

This is different from strings, which are immutable.

## 7. Important List Methods

Python provides several built-in methods to manage lists.

Let's understand them one by one.

### 7.1 append()

Adds one element at the end of the list.

```
fruits = ["Apple", "Banana", "Mango"]
fruits.append("Orange")
print(fruits)
```

Output:

```
['Apple', 'Banana', 'Mango', 'Orange']
```

### 7.2 insert()

Adds an element at a specific position.

Syntax:

```
list_name.insert(index, value)
```

Example:

```
fruits = ["Apple", "Banana", "Mango"]
fruits.insert(1, "Orange")
print(fruits)
```

Output:

```
['Apple', 'Orange', 'Banana', 'Mango']
```

### 7.3 remove()

Removes an element by its value.

```
fruits = ["Apple", "Banana", "Mango"]
fruits.remove("Banana")
print(fruits)
```

Output:

```
['Apple', 'Mango']
```

Important: If duplicate values exist, `remove()` deletes only the first matching occurrence.

### 7.4 pop()

Removes an element using its index and returns the removed element.

```
fruits = ["Apple", "Banana", "Mango"]
removed_fruit = fruits.pop(1)
print(removed_fruit)
print(fruits)
```

Output:

```
Banana
['Apple', 'Mango']
```

If you don't provide an index, `pop()` removes the last element.

```
fruits = ["Apple", "Banana", "Mango"]
fruits.pop()
print(fruits)
```

Output:

```
['Apple', 'Banana']
```

### 7.5 sort()

Sorts the list in ascending order by default.

```
numbers = [50, 10, 40, 20, 30]
numbers.sort()
print(numbers)
```

Output:

```
[10, 20, 30, 40, 50]
```

For descending order:

```
numbers.sort(reverse=True)
print(numbers)
```

Output:

```
[50, 40, 30, 20, 10]
```

### 7.6 reverse()

Reverses the existing order of elements.

```
numbers = [10, 20, 30, 40, 50]
numbers.reverse()
print(numbers)
```

Output:

```
[50, 40, 30, 20, 10]
```

Note: `reverse()` does not sort the list. It simply reverses the current order.

### 7.7 extend()

Adds multiple elements from another iterable to the existing list.

```
fruits = ["Apple", "Banana"]
fruits.extend(["Mango", "Orange"])
print(fruits)
```

Output:

```
['Apple', 'Banana', 'Mango', 'Orange']
```

### 7.8 count()

Counts how many times a value appears.

```
numbers = [10, 20, 10, 30, 10, 40]
print(numbers.count(10))
```

Output:

```
3
```

### 7.9 index()

Returns the index of the first occurrence of a value.

```
fruits = ["Apple", "Banana", "Mango"]
print(fruits.index("Mango"))
```

Output:

```
2
```

### 7.10 clear()

Removes all elements from a list.

```
fruits = ["Apple", "Banana", "Mango"]
fruits.clear()
print(fruits)
```

Output:

```
[]
```

### 7.11 copy()

Creates a shallow copy of a list.

```
numbers = [10, 20, 30]
new_numbers = numbers.copy()
print(new_numbers)
```

Output:

```
[10, 20, 30]
```

### Quick reference – List methods

| Method      | Purpose                            |  Example
| ----------- | ---------------------------------- |------------------------------
| `append()`  | Add one item at the end            | fruits.append("Orange")
| `insert()`  | Add an item at a specific position | fruits.insert(1, "Orange")
| `remove()`  | Remove an item by value            | fruits.remove("Banana")
| `pop()`     | Remove and return an item by index | removed_fruit = fruits.pop(1) -- If no index is provided then last element
| `sort()`    | Sort the list                      | numbers.sort()
| `reverse()` | Reverse the current order          | numbers.reverse()
| `extend()`  | Add multiple items                 | fruits.extend(["Mango", "Orange"])
| `count()`   | Count occurrences                  | numbers.count(10)
| `index()`   | Find the first matching index      | fruits.index("Mango")
| `clear()`   | Remove all items                   | fruits.clear()
| `copy()`    | Create a shallow copy              | new_numbers = numbers.copy()

Important concept: Methods like `append()`, `sort()`, and `reverse()` modify the original list. They generally return `None`, rather than returning the modified list.

# Part 5 – Introduction to Tuples

## 8. What is a Tuple?

A tuple is another type of collection in Python.

Like a list, a tuple can store multiple values.

However, the most important difference is:

A list is mutable, whereas a tuple is immutable.

Tuples are created using parentheses `()`.

Example 1 – Creating a tuple

```
fruits = ("Apple", "Banana", "Mango")
print(fruits)
```

Output:

```
('Apple', 'Banana', 'Mango')
```

Example 2 – Tuple with different data types

```
student = ("Sanket", 25, 85.5, True)
print(student)
```

Output:

```
('Sanket', 25, 85.5, True)
```

Example 3 – Creating a single-element tuple

This is important!

```
numbers = (10,)
print(type(numbers))
```

Output:

```
<class 'tuple'>
```

The comma is necessary.

Without the comma:

```
numbers = (10)
print(type(numbers))
```

Output:

```
<class 'int'>
```

Python considers this an integer, not a tuple.

## 9. Accessing Tuple Elements

Tuples support indexing just like lists.

```
languages = ("Python", "Java", "SQL", "R")
print(languages[0])
print(languages[-1])
```

Output:

```
Python
R
```

## 10. Tuple Slicing

Tuples also support slicing.

```
languages = ("Python", "Java", "SQL", "R", "C++")
print(languages[1:4])
```

Output:

```
('Java', 'SQL', 'R')
```

## 11. Tuple Methods

Tuples have fewer built-in methods than lists because they cannot be modified.

The two main methods are:

count()

```
numbers = (10, 20, 10, 30, 10)
print(numbers.count(10))
```

Output:

```
3
```

index()

```
fruits = ("Apple", "Banana", "Mango")
print(fruits.index("Mango"))
```

Output:

```
2
```

## 12. Why are Tuples Immutable?

Consider this example:

```
fruits = ("Apple", "Banana", "Mango")
fruits[1] = "Orange"
```

Python produces an error:

```
TypeError: 'tuple' object does not support item assignment
```

This happens because tuples do not allow you to modify their elements after creation.

You can still create a new tuple using existing values, but you cannot modify the original tuple in place.

# Part 6 – Lists vs Tuples

## 13. Key Differences

| Feature           | List              | Tuple                          |
| ----------------- | ----------------- | ------------------------------ |
| Syntax            | `[]`              | `()`                           |
| Mutable           | Yes               | No                             |
| Ordered           | Yes               | Yes                            |
| Allows duplicates | Yes               | Yes                            |
| Supports indexing | Yes               | Yes                            |
| Supports slicing  | Yes               | Yes                            |
| Methods           | Many              | Mainly `count()` and `index()` |
| Typical use       | Data that changes | Data that stays fixed          |

### When should you use Lists?

Use lists when:

- You need to add or remove elements.
- You need to update values.
- You are storing data that changes frequently.

Example: Student marks, shopping cart items, employee records.

### When should you use Tuples?

Use tuples when:

- Your collection should remain unchanged.
- You want to represent fixed groups of values.
- You are storing coordinates, dates, or fixed configuration values.

Example:

```
coordinates = (19.0760, 72.8777)
print(coordinates)
```

Output:

```
(19.076, 72.8777)
```
