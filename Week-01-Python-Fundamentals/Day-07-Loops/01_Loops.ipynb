## 1. What are loops?

A loop allows us to execute a block of code repeatedly without writing the same instructions multiple times.

Imagine you want to print:

```
Hello Python
Hello Python
Hello Python
Hello Python
Hello Python
```

Without a loop, you would write `print()` five times.

With a loop, you can achieve the same result using just a few lines of code.

## 2. Types of loops in Python

Python provides two main types of loops:

1. `for` loop
2. `while` loop

| For loop                                           | While loop                                                        |
| -------------------------------------------------- | ----------------------------------------------------------------- |
| Used to iterate over a sequence or iterable.       | Used to repeat code while a condition is True.                    |
| Often used when the number of iterations is known. | Often used when the number of iterations is not known in advance. |

## 3. The `for` loop

A `for` loop is used to execute a block of code for each item in a sequence.

Syntax:

```
for variable in sequence:
    # Code to execute
```

Example:

```
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
    
```

Output:

```
Apple
Banana
Mango
```

How it works:

1. Python takes the first item from the list.
2. Assigns it to the variable `fruit`.
3. Executes the code inside the loop.
4. Moves to the next item.
5. Stops when all items have been processed.

## 4. The `range()` function

The `range()` function generates a sequence of numbers. It is commonly used with `for` loops.

There are three ways to use it.

A. `range(stop)`

Starts at 0 and stops before the specified number.

```
for i in range(5):
    print(i)

```

Output:

```
0
1
2
3
4
```

B. `range(start, stop)`

Starts at the specified number and stops before the ending number.

```
for i in range(1, 6):
    print(i)
    
```

Output:

```
1
2
3
4
5
```

C. `range(start, stop, step)`

The step determines how much the number increases or decreases each time.

```
for i in range(2, 11, 2):
    print(i)

```

Output:

```
2
4
6
8
10
```

You can also count backwards:

```
for i in range(5, 0, -1):
    print(i)

```

Output:

```
5
4
3
2
1
```

Remember: The stop value is never included in `range()`.

## 5. The `while` loop

A `while` loop executes a block of code as long as its condition remains True.

Syntax:

```
while condition:
    # Code to execute
```

Example:

```
count = 1

count = 1

while count <= 5:
    print(count)
    count += 1
```

Output:

```
1
2
3
4
5
```

How it works:

1. `count` starts at 1.
2. Python checks whether `count <= 5`.
3. If True, it prints the value.
4. `count += 1` increases the value by 1.
5. Python checks the condition again.
6. When `count` becomes 6, the condition is False and the loop stops.


Important : Always make sure your `while` loop has a way to eventually become False. Otherwise, you may create an infinite loop.



## 6. Infinite loops

An infinite loop continues running because its condition never becomes False.

Example:

```
count = 1

while count <= 5:
    print(count)
```

Here, `count` never increases, so the condition remains True forever.

The corrected version is:

```
count = 1

while count <= 5:
    print(count)    
    count += 1
```

## 7. The `break` statement

The `break` statement immediately terminates a loop.

Example:

```
for i in range(1, 10):
    if i == 5:
        break

    print(i)
```

Output:

```
1
2
3
4
```

When `i` becomes 5, the loop stops completely.

## 8. The `continue` statement

The `continue` statement skips the current iteration and moves to the next iteration.

Example:

```
for i in range(1, 6):
    if i == 3:
        continue

    print(i)
```

Output:

```
1
2
4
5
```

The number 3 is skipped, but the loop continues.

## 9. The `else` clause with loops

Python allows us to use `else` with both `for` and `while` loops.

The `else` block executes when the loop finishes normally, without encountering `break`.

Example:

```
for i in range(1, 4):
    print(i)
else:
    print("Loop completed")
    
```

Output:

```
1
2
3
Loop completed
```

Example with `break`:

```
for i in range(1, 5):
    if i == 3:
        break

    print(i)
else:
    print("Loop completed")
```

Output:

```
1
2
```

The `else` block does not execute because the loop ended with `break`.

## 10. Nested loops

A nested loop means placing one loop inside another loop.

Example:

```
for i in range(1, 4):
    for j in range(1, 3):
        print(i, j)
```

Output:

```
1 1
1 2
2 1
2 2
3 1
3 2
```

The inner loop completes all its iterations for each iteration of the outer loop.

## 11. Useful loop patterns

A. Calculate the sum of numbers

```
total = 0

for i in range(1, 6):
    total += i

print("Total:", total)
```

Output:

```
Total: 15
```

B. Iterate over a string

```
name = "Python"

for character in name:
    print(character)

```

Output:

```
P
y
t
h
o
n
```

C. Iterate over a dictionary

```
student = {
    "name": "Sanket",
    "age": 25,
    "subject": "Python"
}

for key, value in student.items():
    print(key, ":", value)
```

Output:

```
name : Sanket
age : 25
subject : Python
```

## 12. `for` vs `while` – Quick comparison

| Feature            | `for` loop             | `while` loop                           |
| ------------------ | ---------------------- | -------------------------------------- |
| Main purpose       | Iterate over items     | Repeat while condition is True         |
| Condition          | Managed by iterable    | Explicit condition                     |
| Counter            | Often automatic        | Often manually updated                 |
| Infinite loop risk | Lower                  | Higher                                 |
| Common use         | Lists, strings, ranges | User input, condition-based repetition |


## Exercise 12 – Number Guessing Game

Create a simple number guessing game.

- Store a secret number in a variable.
- Ask the user to enter a number using `input()`.
- If the guess is correct, print `"Correct Guess!"`.
- Otherwise, print `"Wrong Guess!"`.

Example:

```
secret_number = 7
```

Hint: Convert the input into an integer using `int(input())`. You can use a `while` loop to allow repeated guesses as an additional challenge.


````python
secret_number = 7
guessed_number = 0

while guessed_number != secret_number:
    guessed_number = int(input("Guess the number: "))

    if guessed_number == secret_number:
        print("Correct Guess!")
    else:
        print("Wrong Guess! Try again.")
````
