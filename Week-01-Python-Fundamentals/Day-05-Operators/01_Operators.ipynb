## 1. Introduction to Operators

### What are operators?

Operators are special symbols or keywords in Python that perform operations on values and variables.

For example:

```
a = 10
b = 5
result = a + b
print(result)
```

Output:

```
15
```

Here:

- `a` and `b` are variables.
- `10` and `5` are values.
- `+` is an operator.
- `a` and `b` are called operands.
- `a + b` is an expression.

### What are operands?

Operands are the values or variables on which an operator performs an operation.

Example:

```
number1 = 20
number2 = 10
result = number1 - number2
print(result)
```

In this example:

| Element             | Description    |
| ------------------- | -------------- |
| `number1`           | First operand  |
| `-`                 | Operator       |
| `number2`           | Second operand |
| `number1 - number2` | Expression     |

### Types of Python Operators

Python has seven main categories of operators:

| No. | Operator category | Purpose                           |
| --- | ----------------- | --------------------------------- |
| 1   | Arithmetic        | Mathematical calculations         |
| 2   | Comparison        | Compare two values                |
| 3   | Assignment        | Assign or update values           |
| 4   | Logical           | Combine conditions                |
| 5   | Identity          | Check object identity             |
| 6   | Membership        | Check whether an item exists      |
| 7   | Bitwise           | Perform operations on binary bits |

We'll cover all seven categories, followed by operator precedence and practical exercises.

## 2. Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations.

| Operator | Name           | Example   | Result |
| -------- | -------------- | --------- | ------ |
| `+`      | Addition       | `10 + 5`  | 15     |
| `-`      | Subtraction    | `10 - 5`  | 5      |
| `*`      | Multiplication | `10 * 5`  | 50     |
| `/`      | Division       | `10 / 5`  | 2.0    |
| `//`     | Floor Division | `10 // 3` | 3      |
| `%`      | Modulus        | `10 % 3`  | 1      |
| `**`     | Exponentiation | `2 ** 3`  | 8      |

### Example 1 – Addition

```
a = 10
b = 5
print(a + b)
```

Output:

```
15
```

### Example 2 – Division

```
a = 10
b = 3
print(a / b)
```

Output:

```
3.3333333333333335
```

Remember: `/` always performs regular division and returns a floating-point result.

### Example 3 – Floor Division (`//`)

Floor division returns the quotient rounded down to the nearest integer (or whole-number value).

```
a = 10
b = 3
print(a // b)
```

Output:

```
3
```

### Example 4 – Modulus (`%`)

Modulus returns the remainder after division.

```
a = 10
b = 3
print(a % b)
```

Output:

```
1
```

Explanation:

- 10 divided by 3 gives quotient 3.
- 3 × 3 = 9.
- 10 − 9 = 1 (remainder).

### Example 5 – Exponentiation (`**`)

Exponentiation calculates the power of a number.

```
a = 2
b = 3
print(a ** b)
```

Output:

```
8
```

Explanation: \\(2^3 = 2 \times 2 \times 2 = 8\\).

### Important practical examples

Check whether a number is even or odd:

```
number = 10

```

Output:

```
0
```

If the remainder is `0`, the number is even. Otherwise, it is odd.

Calculate the last digit of a number:

```

Output:

```

## 3. Comparison Operators

## 5. Logical Operators

Logical operators are used to combine multiple conditions.

Python has three logical operators:

| Operator | Meaning                                  |
| -------- | ---------------------------------------- |
| `and`    | True when both conditions are True       |
| `or`     | True when at least one condition is True |
| `not`    | Reverses the Boolean result              |

### Example 1 – AND operator

Both conditions must be true.

```
age = 25
print(age >= 18 and age <= 60)
```

Output:

```
True
```

### Example 2 – OR operator

At least one condition must be true.

```
day = "Sunday"
print(day == "Saturday" or day == "Sunday")
```

Output:

```
True
```

### Example 3 – NOT operator

Reverses the result.

```
is_logged_in = True
print(not is_logged_in)
```

Output:

```
False
```

### Logical operators – Truth table

| A     | B     | A and B | A or B |
| ----- | ----- | ------- | ------ |
| True  | True  | True    | True   |
| True  | False | False   | True   |
| False | True  | False   | True   |
| False | False | False   | False  |

For `not`:

| A     | not A |
| ----- | ----- |
| True  | False |
| False | True  |

## 6. Identity Operators

Identity operators check whether two variables refer to the same object in memory.

Python has two identity operators:

- `is`
- `is not`

### Example 1 – `is`

```
a = [10, 20, 30]
b = a
print(a is b)
```

Output:

```
True
```

Both variables refer to the same list object.

### Example 2 – `is not`

```
a = [10, 20, 30]
b = [10, 20, 30]
print(a is not b)
```

Output:

```
True
```

Although both lists contain the same values, they are separate objects.

### Important: Difference between `==` and `is`

```
a = [10, 20, 30]
b = [10, 20, 30]
print(a == b)
print(a is b)
```

Output:

```
True
False
```

- `==` checks whether the values are equal.
- `is` checks whether both variables refer to the same object.

## 8. Bitwise Operators

Bitwise operators perform operations on the binary representation of integers.

Binary numbers use only two digits: `0` and `1`.

For example:

| Decimal | Binary |
| ------- | ------ |
| 2       | `0010` |
| 3       | `0011` |
| 5       | `0101` |
| 8       | `1000` |

Python has six bitwise operators:

| Operator | Name        |
| -------- | ----------- |
| `&`      | Bitwise AND |
| `\\|`    | Bitwise OR  |
| `^`      | Bitwise XOR |
| `~`      | Bitwise NOT |
| `<<`     | Left shift  |
| `>>`     | Right shift |

### Example 1 – Bitwise AND (`&`)

Returns 1 only when both corresponding bits are 1.

```
a = 5
b = 3
print(a & b)
```

Output:

```
1
```

Explanation:

```
  0101  (5)
& 0011  (3)
------
  0001  (1)
```

### Example 2 – Bitwise OR (`|`)

Returns 1 when at least one corresponding bit is 1.

```
a = 5b = 3print(a | b)
```

Output:

```
7
```

### Example 3 – Bitwise XOR (`^`)

Returns 1 when the corresponding bits are different.

```
a = 5
b = 3
print(a ^ b)
```

Output:

```
6
```

### Example 4 – Bitwise NOT (`~`)

Inverts the bits of an integer.

```
a = 5
print(~a)
```

Output:

```
-6
```

For Python integers, `~a` is equivalent to `-a - 1`.

### Example 5 – Left shift (`<<`)

Shifts the bits to the left.

```
a = 5
print(a << 1)
```

Output:

```
10
```

### Example 6 – Right shift (`>>`)

Shifts the bits to the right.

```
a = 8
print(a >> 1)
```

Output:

```
4
```

## 9. Operator Precedence

Operator precedence determines which operation Python performs first when an expression contains multiple operators.

Example:

```
result = 10 + 5 * 2
print(result)
```

Output:

```
20
```

Why? Multiplication is performed before addition.

### Basic precedence order

From higher to lower precedence:

| Priority | Operator             |
| -------- | -------------------- |
| 1        | Parentheses `()`     |
| 2        | Exponentiation `**`  |
| 3        | Unary `+`, `-`, `~`  |
| 4        | `*`, `/`, `//`, `%`  |
| 5        | `+`, `-`             |
| 6        | Comparison operators |
| 7        | `not`                |
| 8        | `and`                |
| 9        | `or`                 |

### Example 1 – Using parentheses

```
result = (10 + 5) * 2
print(result)
```

Output:

```
30
```

### Example 2 – Multiple operators

```
result = 10 + 5 * 2 - 4
print(result)
```

Output:

```
16
```

### What is associativity?

Associativity determines the evaluation direction when operators have the same precedence.

For example, most arithmetic operators are evaluated from left to right:

```
result = 20 - 5 - 3
print(result)
```

Output:

```
12
```

Exponentiation is an important exception: it is evaluated from right to left.

```
result = 2 ** 3 ** 2
print(result)
```

Output:

```
512
```

This is equivalent to `2 ** (3 ** 2)`.
