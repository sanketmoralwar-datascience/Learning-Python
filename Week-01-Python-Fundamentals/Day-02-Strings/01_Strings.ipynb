Each character in a string has a position called an index.

Python indexing starts from 0.

Consider this string:

```python
language = "PYTHON"
```

# Positive indexing
P Y T H O N

0 1 2 3 4 5

# Negative indexing
P   Y  T  H  O  N
-6 -5 -4 -3 -2 -1 

Positive indexing starts at 0; negative indexing starts at -1 from the right.
Try this in your Jupyter notebook:

```python
language = "PYTHON"

print(language[0])
print(language[1])
print(language[5])

print(language[-1])
print(language[-2])
```

# Next lesson – String Slicing
String slicing allows you to extract a portion of a string.

The syntax is:
string[start:end:step]

start: Starting index (included).
end: Ending index (excluded).
step: How many positions to move each time.

```python
language = "PYTHON"

print(language[0:3])
print(language[2:5])
print(language[1:])
print(language[:4])
```

# Day 02 – Lesson 3: Python String Methods

# Part 1 – Theory (copy into one Markdown cell)
Python String Methods
String methods are built-in functions in Python that help us manipulate, search, modify and analyse text.

A string method is called using dot notation:

string.method()

For example:

message.upper()

Here, upper() is a string method that converts all lowercase characters into uppercase.
1. Case Conversion Methods
Method	Description	Example
upper()	Converts to uppercase	"hello".upper()
lower()	Converts to lowercase	"HELLO".lower()
title()	Capitalises the first letter of each word	"hello world".title()
capitalize()	Capitalises only the first character	"hello world".capitalize()
swapcase()	Changes uppercase to lowercase and vice versa	"Hello".swapcase()

2. Removing Whitespace
Whitespace includes spaces, tabs and newline characters.
Method	Description
strip()	Removes whitespace from both ends
lstrip()	Removes whitespace from the left
rstrip()	Removes whitespace from the right

Example:

"  Python  ".strip() returns "Python".

Important: These methods do not remove spaces between words.
3. Searching Within Strings
Method	Description
find()	Returns the first position of a substring; returns -1 if not found
index()	Returns the first position; raises an error if not found
count()	Counts how many times a substring appears
startswith()	Checks whether a string starts with specified text
endswith()	Checks whether a string ends with specified text

These methods are useful when searching for specific words or characters.
4. Replacing Text
The replace() method replaces one part of a string with another.

Syntax:

string.replace(old, new)

Example:

"I like Java".replace("Java", "Python")

Result: "I like Python"

You can also specify the maximum number of replacements.
5. Splitting Strings
The split() method divides a string into a list.

Example:

"Python is easy".split()

Result:

["Python", "is", "easy"]

You can also specify a separator:

"apple,banana,mango".split(",")

Result:

["apple", "banana", "mango"]

This is especially useful when processing CSV files and text data.
6. Joining Strings
The join() method combines multiple strings into one string.

Example:

"-".join(["Python", "Java", "C++"])

Result:

"Python-Java-C++"

Remember: split() separates a string, whereas join() combines strings.
7. Checking String Content
Python provides several methods that return Boolean values (True or False).
Method	Description
isalpha()	Checks whether all characters are alphabetic
isdigit()	Checks whether all characters are digits
isalnum()	Checks whether all characters are letters or digits
isspace()	Checks whether all characters are whitespace
islower()	Checks whether all letters are lowercase
isupper()	Checks whether all letters are uppercase
istitle()	Checks whether the string follows title case

Important: isdigit() does not mean that a value is an integer variable. It checks the characters in a string.
8. Important Concept – Strings Are Immutable
Strings in Python are immutable. This means that their contents cannot be changed directly after creation.

For example:

message = "hello"

Calling message.upper() does not change the original variable automatically.

To keep the converted value, assign it to a variable:

message = message.upper()

This concept is important when working with strings in larger Python programs.

# String Formatting (f-strings)

# Part 1 – Theory (copy into one Markdown cell)
Python String Formatting
String formatting allows us to insert variables and expressions into strings.

Python supports several approaches:

1. f-strings (recommended)

An f-string starts with the letter f before the opening quotation mark. Variables and expressions are placed inside curly braces {}.

Example:

name = "Sanket"

print(f"Hello {name}")

Output: Hello Sanket

2. Using format()

The format() method inserts values into placeholders.

Example:

print("Hello {}".format(name))

Output: Hello Sanket

3. Using % formatting

Python also supports older-style formatting using %s, %d and %f.
Placeholder	Meaning
%s	String
%d	Integer
%f	Floating-point number

Example:

print("My age is %d" % 30)

Output: My age is 30

4. Expressions inside f-strings

You can perform calculations directly inside curly braces.

Example:

price = 100

quantity = 3

print(f"Total: {price * quantity}")

Output: Total: 300

5. Formatting decimal places

You can control how many decimal places are displayed.

Example:

price = 99.98765

print(f"Price: {price:.2f}")

Output: Price: 99.99

Here, .2f means display two digits after the decimal point.

# Part 2 – Code examples

```python
# Example 1 – Basic f-strings

name = "Sanket"
age = 30

print(f"My name is {name} and I am {age} years old.")

```

```python
# Example 2 – Expressions

num1 = 100
num2 = 200

print(f"Sum = {num1 + num2}")
print(f"Difference = {num2 - num1}")
print(f"Product = {num1 * num2}")
```

```python
# Example 3 – Decimal formatting

price = 123.45678

print(f"Original price: {price}")
print(f"Formatted price: {price:.2f}")
print(f"Formatted price: {price:.3f}")
```
