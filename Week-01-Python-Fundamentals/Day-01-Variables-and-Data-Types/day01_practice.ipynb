Your exercises 5:
1. Convert num = "250" into an integer.
2. Convert price = 99 into a float.
3. Convert score = 85.5 into an integer.
4. Convert age = 30 into a string.

```python
num = "250"
price = 99
score = 85.5
age = 30

print(f"num type is {type(num)}" )
print(f"price type is {type(price)}")
print(f"score type is {type(score)}")
print(f"age type is {type(age)}\n")

converted_num = int(num)
converted_price = float(price)
converted_score = int(score)
converted_age = str(age)
rounded_score = round(score)

print("After conversion\n")

print(f"num type is {type(converted_num)}" )
print(f"price type is {type(converted_price)}")
print(f"score type is {type(converted_score)}")
print(f"age type is {type(converted_age)}")
print(f"age type is {type(converted_age)}")
print(f"Score: {converted_score}, Rounded_Score = {rounded_score}")
print(f"age type is {type(rounded_score)}")
```
