## 1. Introduction to Conditional Statements

### What are conditional statements?

Conditional statements allow Python to make decisions based on whether a condition is `True` or `False`.

For example:

- If a student scores more than 40 marks, they pass.
- If someone is 18 or older, they are eligible to vote.
- If a number is divisible by 2, it is even; otherwise, it is odd.

In Python, we use conditional statements to implement this kind of decision-making.

### Basic syntax

```
if condition:    # Code executes when condition is True
```

Important rules:

- The `if` keyword starts the condition.
- A colon (`:`) is required after the condition.
- The code inside the condition must be indented.
- Python uses indentation (usually four spaces) to identify code blocks.

### Example

```
age = 25
if age >= 18:
    print("You are eligible to vote")
print("Program completed")
```

Output:

```
You are eligible to vote
Program completed
```

Here, because `age >= 18` is `True`, Python executes the indented statement.

## 2. The `if` Statement

The `if` statement executes a block of code only when the given condition is true.

### Example 1 – Check whether a number is positive

```

if number > 0:
```

Output:

```
Number is positive
```

### Example 2 – Check whether a student passed


Output:

```

```

```
marks = 30
if marks >= 40:
```

Output:

```
```

Nothing is printed because the condition is false.

Remember: An `if` statement does not require an `else` statement.

## 3. The `if-else` Statement

Sometimes we want Python to execute one block when a condition is true and another block when it is false.

For this, we use `if-else`.

### Syntax

```
if condition:    # Executes when condition is True
else:    # Executes when condition is False
```

### Example 1 – Even or Odd

```
number = 7
if number % 2 == 0:
    print("Number is Even")

else:    
    print("Number is Odd")
```

Output:

```
Number is Odd
```

### Example 2 – Pass or Fail

```
marks = 35
if marks >= 40:
    print("You passed")
    
else:
    print("You failed")
```

Output:

```
You failed
```

### Example 3 – Check voting eligibility

```
age = 17

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

Output:

```
Not eligible to vote
```

## 4. The `if-elif-else` Statement

What if we have multiple conditions?

For example, a student receives a grade based on marks:

| Marks    | Grade |
| -------- | ----- |
| 90–100   | A     |
| 80–89    | B     |
| 70–79    | C     |
| 60–69    | D     |
| Below 60 | Fail  |

We can use `if-elif-else` to handle multiple conditions.

### Syntax

```
if condition1:
    # Executes if condition1 is True
elif condition2:
    # Executes if condition2 is True
elif condition3:
    # Executes if condition3 is True
else:
    # Executes if all conditions are False
```

`elif` means "else if".

### Example 1 – Student grading system

```
marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else: 
    print("Fail")
```

Output:

```
Grade B
```

### How does Python evaluate this?

1. Checks whether marks are greater than or equal to 90. False.
2. Checks whether marks are greater than or equal to 80. True.
3. Prints Grade B.
4. Skips all remaining conditions.

Important: Python stops checking once it finds the first true condition in an `if-elif-else` chain.

### Example 2 – Positive, Negative, or Zero

```
number = -5

if number > 0:
    print("Positive number")
elif number < 0:
    print("Negative number")
else:
    print("Number is zero")
```

Output:

```
Negative number
```

## 5. Nested `if` Statements

A nested `if` means placing one `if` statement inside another `if` statement.

We use nested conditions when a decision depends on another decision.

### Syntax

```
if condition1:
    if condition2:
        # Executes when both conditions are True
    else:
        # Executes when condition1 is True
        # but condition2 is False
else:
    # Executes when condition1 is False
```

### Example – Login system

```
username = "admin"
password = "12345"

if username == "admin":
    if password == "12345":
        print("Login successful")
    else:
        print("Incorrect password")
else:
    print("Invalid username")
```

Output:

```
Login successful
```

### Example – Check eligibility for a discount

```
age = 65
membership = True

if age >= 60:
    if membership:
        print("You are eligible for a special discount")
    else:
        print("Please register for membership")
else:
    print("You are not eligible for this discount")
```

Output:

```
You are eligible for a special discount
```

Remember: Indentation is especially important in nested conditions.

## 6. Using Logical Operators with Conditional Statements

In Day 05, we learned three logical operators:

- `and`
- `or`
- `not`

Now let's use them with conditional statements.

### Example 1 – Using AND

Both conditions must be true.

```
age = 25
salary = 50000

if age >= 18 and salary >= 30000:
    print("Eligible for loan")
else:
    print("Not eligible for loan")
```

Output:

```
Eligible for loan
```

### Example 2 – Using OR

At least one condition must be true.

```
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("It's a weekend")
else:
    print("It's a working day")
```

Output:

```
It's a weekend
```

### Example 3 – Using NOT

Reverses the Boolean condition.

```
is_raining = False

if not is_raining:
    print("You can go outside")
else:
    print("Take an umbrella")
```

Output:

```
You can go outside
```

## 7. Short-hand Conditional Expression (Ternary Operator)

Python allows us to write a simple `if-else` statement in one line.

This is called a conditional expression or ternary operator.

### Syntax

```
value_if_true if condition else value_if_false
```

### Example 1 – Even or Odd

```
number = 10

result = "Even" if number % 2 == 0 else "Odd"

print(result)
```

Output:

```
Even
```

### Example 2 – Pass or Fail

```
marks = 75

result = "Pass" if marks >= 40 else "Fail"

print(result)
```

Output:

```
Pass
```

Note: Use ternary expressions for simple decisions. For complex logic, regular `if-else` statements are easier to read.

## 8. The `pass` Statement

Sometimes you want to write an `if` statement but don't want to implement its logic yet.

Python does not allow an empty block.

We can use the `pass` statement as a placeholder.

### Example

```
age = 25

if age >= 18:
    pass
else:
    print("Not eligible")
```

The `pass` statement does nothing. It simply allows the program to run without an error.

## 9. Important Concepts and Common Mistakes

### Mistake 1 – Missing colon

Incorrect:

```
if age >= 18
    print("Eligible")
```

Correct:

```
if age >= 18:
    print("Eligible")
```

### Mistake 2 – Incorrect indentation

Incorrect:

```
if age >= 18:
print("Eligible")
```

Correct:

```
if age >= 18:
    print("Eligible")
```

### Mistake 3 – Using `=` instead of `==`

Incorrect:

```
if age = 18:
    print("Age is 18")
```

Correct:

```
if age == 18:
    print("Age is 18")
```

### Mistake 4 – Incorrect condition order

Consider this:

```
marks = 95

if marks >= 60:
    print("Grade D")
elif marks >= 90:
    print("Grade A")
```

Output:

```
Grade D
```

Why? Because 95 is already greater than 60, so Python executes the first condition and skips the second.

Correct order:

```
marks = 95

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade D")
```

Output:

```
Grade A
```


## Exercise 12 – Mini Challenge: Electricity Bill Calculator

Create a variable:

```
units = 250
```

Calculate the electricity bill using these rates:

| Units consumed  | Rate         |
| --------------- | ------------ |
| First 100 units | ₹5 per unit  |
| Next 100 units  | ₹7 per unit  |
| Above 200 units | ₹10 per unit |

For example, if consumption is 250 units:

- First 100 units: 100 × ₹5
- Next 100 units: 100 × ₹7
- Remaining 50 units: 50 × ₹10

Calculate the total bill using conditional statements.

Hint: You will need `if-elif-else` and some arithmetic calculations.


````python
units = 250

if units > 200:
    total_bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)
elif units > 100:
    total_bill = (100 * 5) + ((units - 100) * 7)
else:
    total_bill = units * 5

print(total_bill)
````
