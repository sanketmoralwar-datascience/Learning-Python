# Part 1 – Dictionaries


1. What is a Dictionary in Python?
A dictionary is a collection that stores information in key-value pairs.

Let's understand this with a real-world example.

Imagine you want to store information about a student:
- Name: Sanket
- Age: 30
- City: Thane
- Course: Python
You could store these details in separate variables:
name = "Sanket"age = 30city = "Thane"course = "Python"



But Python provides a better way to organize related information using dictionaries.
student = {    "name": "Sanket",    "age": 30,    "city": "Thane",    "course": "Python"}
print(student)



Output:
{'name': 'Sanket', 'age': 30, 'city': 'Thane', 'course': 'Python'}


Here:
Key	Value
name	Sanket
age	30
city	Thane
course	Python

Think of a dictionary like a contact book.

You search for a person's name (key) to get their phone number (value).
Important characteristics of Dictionaries
1. Key-value pairs: Every entry contains a key and a value.
2. Ordered: Dictionaries maintain insertion order.
3. Mutable: You can add, update, and remove elements.
4. Unique keys: Duplicate keys are not allowed. A later value replaces an earlier value for the same key.
5. Different data types: Values can contain strings, integers, lists, tuples, and even other dictionaries.
6. Keys must be hashable: Strings, integers, and tuples containing hashable elements can be used as keys. Lists and dictionaries cannot be dictionary keys.


2. Creating Dictionaries
   
Dictionaries are created using curly brackets {}.

Example 1 – Dictionary of student details
student = {    "name": "Sanket",    "age": 30,    "marks": 85,    "passed": True}
print(student)



Output:
{'name': 'Sanket', 'age': 30, 'marks': 85, 'passed': True}


Example 2 – Empty dictionary
student = {}
print(student)
print(type(student))



Output:
{}
<class 'dict'>


Example 3 – Using the dict() constructor
student = dict(name="Sanket", age=30, city="Thane")
print(student)



Output:
{'name': 'Sanket', 'age': 30, 'city': 'Thane'}


Example 4 – Dictionary with a list as a value
student = {    "name": "Sanket",    "subjects": ["Python", "SQL", "Statistics"]}
print(student)



Output:
{'name': 'Sanket', 'subjects': ['Python', 'SQL', 'Statistics']}


Notice that a dictionary can store a list as its value.
3. Accessing Dictionary Values
Unlike lists, dictionaries use keys rather than numerical indexes to access values.

Example 1 – Access using square brackets
student = {    "name": "Sanket",    "age": 30,    "city": "Thane"}
print(student["name"])
print(student["age"])



Output:
Sanket
30


Example 2 – Access using get()

Python also provides the get() method.
student = {    "name": "Sanket",    "age": 30,    "city": "Thane"}
print(student.get("city"))
print(student.get("course"))



Output:
Thane
None


What's the difference?
Method	If key exists	If key doesn't exist
student["name"]	Returns value	Raises KeyError
student.get("name")	Returns value	Returns None by default

You can also provide a default value:
print(student.get("course", "Not Available"))



Output:
Not Available


Remember: get() is useful when you're not sure whether a key exists.
4. Adding and Updating Dictionary Elements
Dictionaries are mutable, which means you can modify them after creation.

Example 1 – Adding a new key-value pair
student = {    "name": "Sanket",    "age": 30}
student["city"] = "Thane"
print(student)



Output:
{'name': 'Sanket', 'age': 30, 'city': 'Thane'}


Example 2 – Updating an existing value
student = {    "name": "Sanket",    "age": 30}
student["age"] = 31
print(student)



Output:
{'name': 'Sanket', 'age': 31}


Example 3 – Using update()

The update() method can add multiple key-value pairs or update existing ones.
student = {    "name": "Sanket",    "age": 30}
student.update({    "age": 31,    "city": "Thane",    "course": "Python"})
print(student)



Output:
{'name': 'Sanket', 'age': 31, 'city': 'Thane', 'course': 'Python'}


5. Removing Dictionary Elements
Python provides several methods to remove dictionary elements.
5.1 pop()
Removes a specific key and returns its value.
student = {    "name": "Sanket",    "age": 30,    "city": "Thane"}
removed_value = student.pop("age")
print(removed_value)
print(student)



Output:
30
{'name': 'Sanket', 'city': 'Thane'}


5.2 popitem()
Removes and returns the last inserted key-value pair.
student = {    "name": "Sanket",    "age": 30,    "city": "Thane"}
student.popitem()
print(student)



Output:
{'name': 'Sanket', 'age': 30}


5.3 del
Deletes a specific key-value pair.
student = {    "name": "Sanket",    "age": 30,    "city": "Thane"}
del student["city"]
print(student)



Output:
{'name': 'Sanket', 'age': 30}


5.4 clear()
Removes all dictionary elements.
student = {    "name": "Sanket",    "age": 30}
student.clear()
print(student)



Output:
{}


6. Important Dictionary Methods
Let's learn some additional methods.

keys() – Returns all dictionary keys.
student = {    "name": "Sanket",    "age": 30,    "city": "Thane"}
print(student.keys())



Output:
dict_keys(['name', 'age', 'city'])


values() – Returns all dictionary values.
print(student.values())



Output:
dict_values(['Sanket', 30, 'Thane'])


items() – Returns all key-value pairs.
print(student.items())



Output:
dict_items([('name', 'Sanket'), ('age', 30), ('city', 'Thane')])


copy() – Creates a shallow copy.
new_student = student.copy()
print(new_student)



Output:
{'name': 'Sanket', 'age': 30, 'city': 'Thane'}


Dictionary methods – Quick reference
Method	Purpose
get()	Access a value safely
keys()	Get all keys
values()	Get all values
items()	Get key-value pairs
update()	Add or update elements
pop()	Remove a specific key
popitem()	Remove the last inserted pair
clear()	Remove all elements
copy()	Create a shallow copy

7. Nested Dictionaries
A nested dictionary is a dictionary inside another dictionary.

This is useful when you want to store information about multiple students.

Example:
students = {    "student1": {        "name": "Sanket",        "age": 30,        "marks": 85    },    
                "student2": {        "name": "Rahul",        "age": 28,        "marks": 90    }}
print(students)



Accessing nested values:
print(students["student1"]["name"])
print(students["student2"]["marks"])



Output:
Sanket
90


Here, we first access the student record and then access a particular key inside it.

# Part 2 – Sets




8. What is a Set in Python?
A set is a collection of unique elements.

Let's understand this with an example.

Imagine you have the following list of subjects:
subjects = ["Python", "SQL", "Python", "Excel", "SQL"]
print(subjects)



Output:
['Python', 'SQL', 'Python', 'Excel', 'SQL']


Notice that Python and SQL appear more than once.

If you want only unique values, you can use a set.
subjects = {"Python", "SQL", "Python", "Excel", "SQL"}
print(subjects)



Output:
{'Python', 'SQL', 'Excel'}


The order of set elements is not guaranteed, so your output may display them in a different order.
Important characteristics of Sets
1. Unique elements: Duplicate values are automatically removed.
2. Unordered: Sets do not maintain a guaranteed element position.
3. Mutable: You can add and remove elements.
4. No indexing: You cannot access elements using positions like my_set[0].
5. Different data types: Sets can contain multiple hashable data types.
6. Immutable elements: Individual set elements must be hashable. Lists and dictionaries cannot be set elements.
9. Creating Sets
Sets are created using curly brackets {}.

Example 1 – Set of numbers
numbers = {10, 20, 30, 40, 50}
print(numbers)



Output (order may vary):
{10, 20, 30, 40, 50}


Example 2 – Set of strings
languages = {"Python", "Java", "SQL", "R"}
print(languages)



Example 3 – Empty set

This is important!
my_set = set()
print(type(my_set))



Output:
<class 'set'>


Why not use {}?

Because {} creates an empty dictionary, not an empty set.

Example 4 – Convert a list into a set
numbers = [10, 20, 10, 30, 20, 40]
unique_numbers = set(numbers)
print(unique_numbers)



Output (order may vary):
{10, 20, 30, 40}


This is a very useful way to remove duplicates from a list.

10. Adding and Removing Set Elements
10.1 add()
Adds one element to a set.
subjects = {"Python", "SQL"}
subjects.add("Excel")
print(subjects)



Output (order may vary):
{'Python', 'SQL', 'Excel'}


10.2 update()
Adds multiple elements.
subjects = {"Python", "SQL"}

10.3 remove()
Removes a specific element.
subjects = {"Python", "SQL", "Excel"}
subjects.remove("SQL")

Removes an element if it exists.
subjects = {"Python", "SQL", "Excel"}
subjects.discard("Java")
print(subjects)

No error occurs even though Java is not present.

10.5 pop()
Removes and returns an arbitrary element from the set.
subjects = {"Python", "SQL", "Excel"}
removed = subjects.pop()
print(removed)
print(subjects)

Remember: Sets have no guaranteed order, so you cannot predict which element will be removed.


10.6 clear()

11. Set Operations
This is one of the most important parts of today's lesson.

Set operations are useful when comparing two groups of data.

Let's consider two sets:
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}




11.1 Union
Union combines all unique elements from both sets.

Syntax:
A.union(B)



Example:
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A.union(B))



Output (order may vary):
{1, 2, 3, 4, 5, 6, 7, 8}


You can also use the | operator.
print(A | B)



11.2 Intersection
Intersection returns only the common elements between two sets.

Example:
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A.intersection(B))



Output:
{4, 5}


You can also use &.
print(A & B)



11.3 Difference
Difference returns elements that exist in the first set but not in the second set.

Example:
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A.difference(B))



Output:
{1, 2, 3}


Using the - operator:
print(A - B)



Important:
print(B - A)



Output:
{6, 7, 8}


The order of sets matters when using difference.


11.4 Symmetric Difference
Returns elements that are present in either set, but not in both.

Example:
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A.symmetric_difference(B))



Output:
{1, 2, 3, 6, 7, 8}


Using the ^ operator:
print(A ^ B)



Set operations – Quick reference
Operation	Method	Operator
Union	union()	\|
Intersection	intersection()	&
Difference	difference()	-
Symmetric Difference	symmetric_difference()	^

12. Other Useful Set Methods
issubset()

Checks whether all elements of one set are present in another set.
A = {1, 2}
B = {1, 2, 3, 4, 5}
print(A.issubset(B))



Output:
True


issuperset()

Checks whether one set contains all elements of another set.
print(B.issuperset(A))



Output:
True


isdisjoint()

Checks whether two sets have no common elements.
A = {1, 2, 3}
B = {4, 5, 6}
print(A.isdisjoint(B))



Output:
True

Both use curly brackets, but they serve different purposes.
Feature	|Dictionary     |	Set
Syntax	|{key: value}   |   {value1, value2}

Duplicates|	Keys must be unique|	Elements must be unique
Ordered|	Yes, insertion order|	No guaranteed order
Indexing|	Access by key|	No indexing
Mutable|	Yes	| Yes
Empty creation|	{} or dict()|	set()
